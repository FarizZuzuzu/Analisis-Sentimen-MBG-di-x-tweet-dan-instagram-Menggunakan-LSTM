jalankan actor

```python
 # ===================================================================================
# 3. Menjalankan Actor
# ===================================================================================

console.rule("[bold yellow]  Menjalankan Actor Instagram Comment Scraper")
console.print("Harap tunggu, proses scraping sedang berjalan...")

# Remove duplicate URLs from run_input['directUrls'] to prevent InvalidRequestError
run_input['postUrls'] = list(set(run_input['postUrls']))

run = client.actor("clappi/instagram-comments-scraper").call(run_input=run_input)

console.print("[bold green] Actor selesai dijalankan!")
```

ambil hasil Scraping

```python
# =======================================================================
# 4. Mengambil hasil scraping dengan progress bar
#========================================================================

# Setelah Actor selesai, hasilnya disimpan ke Dataset Apify.
# Kita ambil dataset tersebut dan masukkan ke dalam list 'results'.
console.rule("[bold cyan] Mengambil Data dari Dataset")

results = []

# 'run' is already the Run Pydantic model, so its details are directly accessible
# No need to call .get_sync() as 'run' itself holds the details.

dataset_id = run.default_dataset_id

# Auto-detect jumlah item Jika tersedia, Jika tidak pakai estimasi
# → progress bar akan menampilkan estimasi jumlah item
# Access 'item_count' directly from the 'run' object
estimated_total = run.item_count if hasattr(run, 'item_count') and run.item_count is not None else 3000

# Gunakan tqdm untuk menampilkan progress bar saat data diambil
for item in tqdm(client.dataset(dataset_id).iterate_items(),
                 desc="Mengambil data",
                 unit="item",
                 total=estimated_total):
    results.append(item)
```

```python
# ===================================================================================
# 5. Menyimpan ke CSV & Excel
# ===================================================================================

# Penyimpanan data dilakukan setelah scraping selesai.
# File akan disimpan dalam dua format: CSV dan Excel (.xlsx)
console.rule("[bold green]* Menyimpan Data ke File")

# Simpan ke RAW_DATA_PATH di Google Drive
csv_file = RAW_DATA_PATH
excel_file = f"{DRIVE_PATH}/hasil_ig_comments.xlsx"

if results:
  df = pd.DataFrame(results)   # mengubah list hasil scraping jadi dataframe

  # Simpan CSV
  df.to_csv(csv_file, index=False, encoding='utf-8-sig')

  # Simpan Excel menggunakan openpyxl sebagai engine
  df.to_excel(excel_file, index=False, engine="openpyxl")

  # Membuat tabel ringkasan menggunakan rich
  table = Table(title="Ringkasan Output", box=box.HEAVY)
  table.add_column("Jenis File", style="cyan" , justify="center")
  table.add_column ("Nama File", style="green")
  table.add_column ("Jumlah Baris", style="yellow", justify="center")

  table.add_row("CSV", csv_file, str(len(df)))
  table.add_row("Excel", excel_file, str(len(df)))

  console.print (table)

else:
  console.print("[bold red] Tidak ada data ditemukan!")
```
